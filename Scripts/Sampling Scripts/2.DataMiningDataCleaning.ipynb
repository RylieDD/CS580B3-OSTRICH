```python
import pandas as pd
import matplotlib
```

```python
data = pd.read_csv("Full_Community_MSR.csv",usecols = ['PR Number', 'Author', 'Created Time', 'Merged','Time to Merge (hours)','Maintainer Comments'])
```

```python
print(data.head)
print(data.size)
```

```python
data = data[data['Maintainer Comments'].apply(lambda x: x != "[]")]
print(data.size)
print(data.head)
```

```python
data_sorted = data.sort_values("Author")
data_sorted.to_csv("sorted_msr_data.csv")
```

```python
data_grouped = data.groupby('Author')['PR Number']
data_grouped_count = data_grouped.count().plot(kind='bar')
print(data_grouped_count)
```
