```python
import json

def show_every_extension(data, column):
	paths = []
	for row in data:
		if not row[column]:
			print("Received empty column")
			return
		try:
			path_list = json.loads(row[column].replace("'", '"'))
			if isinstance(path_list, list):
				for path in path_list:
					if not isinstance(path, str):
						print(f"Encountered something not a string {data[0]}|/n {path}, \n{paths}")
						return
				paths.extend(path_list)
			else:
				print("Encountered something not a list", path_list)
		except Exception as ex:
			print(f"Failed on |{row[column]}|, {ex}")
			return

	extensions = set()
	for path in paths:
		if "." not in path:
			continue
		decimal_location = path.rindex(".")
		extension = path[decimal_location:]
		if extension == ".template" or extension == ".example" or extension == ".cur" or extension == ".yml" or extension == ".yaml":
			print(path)
		extensions.add(extension)
	return extensions
```

```python
ONE_TENTH = 1.0/10.0

weightings = {'.csv': ONE_TENTH,
 '.cur': 0,
 '.editorconfig': 0,
 '.example': 1,
 '.git-blame-ignore-revs': 1,
 '.gitignore': 1,
 '.json': 0,
 '.md': ONE_TENTH,
 '.png': 0,
 '.py': 1,
 '.snippet': 1,
 '.talon': 1,
 '.talon-list': ONE_TENTH,
 '.template': ONE_TENTH,
 '.txt': ONE_TENTH,
 '.yaml': 1,
 '.yml': 1}
```

```python
def compute_weighted_count(files, added, deleted, weightings):
	result = 0
	for file, a, d in zip(files, added, deleted):
		try:
			decimal_location = file.rindex(".")
		except ValueError as ex:
			print(file, ex)
			return 0
		extension = file[decimal_location:]
		weight = weightings[extension]
		result += weight*(a + d)
	return result
```

```python
def compute_list(text):
	return json.loads(text.replace("'", '"'))

def append_change_sizes(data, add_column, subtract_column, files_column):
	for row in data:
		count = compute_weighted_count(
			compute_list(row[files_column]),
			compute_list(row[add_column]),
			compute_list(row[subtract_column]),
			weightings
		)
		row.append(count)
```

```python
def get_sorted_change_sizes(data):
	return sorted([d[-1] for d in data])
```

```python
def perform_quarter_split(data, lower, mid, upper):
	lowest = []
	low = []
	middle = []
	high = []
	for row in data:
		count = row[-1]
		number = row[0]
		if count <= lower:
			lowest.append(number)
		elif count <= mid:
			low.append(number)
		elif count <= upper:
			middle.append(number)
		else:
			high.append(number)
	return lowest, low, middle, high
```

```python
lowest, low, middle, high = perform_quarter_split(new_data, 6, 24, 101.5)
```

```python
def output_file(lines, path):
	with open(path, "w") as f:
		for line in lines:
			f.write(f"{line}\n")
```

```python
def compute_number_author_pairs(data, numbers):
	needed = set(numbers)
	result = []
	for row in data:
		if row[0] in needed:
			result.append((row[0], row[1]))
	return result
```

```python
import random

def sample(n_train, n_validate, n_test, sources):
	sources = sources[:]
	weightings = compute_weightings(sources)
	train = extract_n_from_multiple_sources(sources, weightings, n_train)
	validate = extract_n_from_multiple_sources(sources, weightings, n_validate)
	test = extract_n_from_multiple_sources(sources, weightings, n_test)
	return train, validate, test

def extract_n_from_multiple_sources(sources, weightings, number):
	result = []
	for source, weighting in zip(sources, weightings):
		extract_n(source, weighting, number, result)
	return result

def compute_weightings(sources):
	result = []
	for source in sources:
		counts_per_author = {}
		for _, author in source:
			if author in counts_per_author:
				counts_per_author[author] += 1
			else:
				counts_per_author[author] = 1
		weightings = []
		for _, author in source:
			count = counts_per_author[author]
			weightings.append(1/count)
		result.append(weightings)
	return result

def extract_n(source, weightings, number, result):
	for _ in range(number):
		value = random.choices(source, weightings, k=1)[0]
		result.append(value)
		i = source.index(value)
		source.pop(i)
		weightings.pop(i)
```

```python
train, validate, test = sample(25, 25, 25, sources)
```
